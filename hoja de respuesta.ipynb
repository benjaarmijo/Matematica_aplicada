```python
#Determinar si un determinado valor es término de la sucesión. (CODIGO)  Y POSICION DEL TERMINO
final=34112
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino=533*2**(n-1)
    dif=final-termino

if dif==0:
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```

```python
#Determinar si un determinado valor es término de la sucesión. (CODIGO)  Y POSICION DEL TERMINO
final=1828000
termino=0
n=0
dif=final-termino
while dif>0:
    n +=1
    termino=1000000+(n-1)*36000
    dif=final-termino

if dif==0:
    print(f'El {final} es un término de la sucesión y ocupa el lugar {n}')

else:
    print(f'El {final} no pertenece a la sucesión')
```
