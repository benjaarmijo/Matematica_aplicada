```python
# 2.

def V(t):
    return 0.85*t**2+13.5*t+38

def S(t):
    return 1.2*t**2+16.2*t+29

def C(t):
    return 1.4*t**2+18.7*t+27



t = np.arange(0, 25, 1)

plt.plot(t, V(t), label='V(t)')   # Grafica la primera función
plt.plot(t, S(t), label='S(t)')   # Grafica la segunda función
plt.plot(t, C(t), label='C(t)')  
plt.plot(15,V(15),marker="o")
plt.plot(15,S(15),marker="o")
plt.plot(15,C(15),marker="o")
plt.plot(17,V(17),marker="o")
plt.plot(17,S(17),marker="o")
plt.plot(17,C(17),marker="o")


plt.title('Relacion entre el acumulamiento de agua y las horas transcurridas')
plt.ylabel('Acumulamiento de agua (milimetros)')
plt.xlabel('Tiempo (horas)')

plt.legend()   # Muestra cuál curva corresponde a f(x) y cuál a g(x)
plt.grid()
plt.show()
```
