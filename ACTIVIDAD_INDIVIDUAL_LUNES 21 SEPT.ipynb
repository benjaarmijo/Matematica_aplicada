# Enunciado

En un experimento de biología se estudia el crecimiento de dos tipos de plantas, durante un período de 9 a 80 días.

La Planta A presenta un crecimiento aproximadamente constante. Por cada 2 días, su altura aumenta 0,8 cm, como se muestra en la tabla.

La altura de la Planta B, en cambio, varía de acuerdo con un modelo cuadrático.

#### Método A:
  $$f(x) = 0,4*x +3,4$$

#### Método B:


$$g(x)=0,01x^2-0,3x+9,7$$

Con ayuda de Python resuelve las siguientes preguntas, considera un decimal para responder:

A_¿Qué tipo de función permite representar el crecimiento de la Planta A? interprete la PENDIENTE


B_Calcule la altura de ambas plantas cuando han transcurrido 27 días. ¿Cuál planta presenta mayor altura?


C_¿En qué día(s) ambas plantas alcanzan la misma altura? ¿Cuál es esa altura? 


D¿Durante qué período la Planta A presenta una altura mayor que la Planta B? (Recuerda GRAFICAR)





# Desarrollo

**1**

El tipo de funcion que representa el crecimiento de la planta A es lineal

por cada dia el crecimiento de la planta A aumenta en 0,4 cm su altura

```python
2.
def f(x):
    return 0.4*x + 3.4

def g(x):
    return 0.01*x**2 - 0.3*x + 9.7

x = 27
print(f"la altura de la planta a al transcurrir 27 dias es de {f(x):.2f}")
print(f"la altura de la planta b al transcurrir 27 dias es de {g(x):.2f}")
```

```python
3.
def T(x):
  return f(x)-g(x)

x = np.linspace(0, 80, 10)  # crea puntos de partida  para  para que solve busque las respuestas

#np.around() permite redondear los valores a 2 decimales y elimina los restantes
solucion = np.around(fsolve(T, x), decimals = 0)

#np.unique() permite eliminar las soluciones repetidas
interseccion = np.unique(solucion)

print(f'Los dias en los que las plantas alcanzan la misma altura son {interseccion}')    # cuando son varias

#print(f'Los puntos de intersección son x1 = {interseccion[0]} y x2 = {interseccion[1]}')  #  cuando son dos
```

```python
4.
def f(x):
    return 0.4*x + 3.4

def g(x):
    return 0.01*x**2 - 0.3*x + 9.7


x = np.arange(0, 100, 0.1)

plt.plot(x, f(x), label='Planta A')   # Grafica la primera función
plt.plot(x, g(x), label='Planta B')   # Grafica la segunda función

plt.title('Funcion del crecimiento en dias')
plt.ylabel('Crecimiento (cm)')
plt.xlabel('Tiempo (Dias)')

plt.legend()   # Muestra cuál curva corresponde a f(x) y cuál a g(x)
plt.grid()
plt.show()
```
